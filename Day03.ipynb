```python
# POS Mapping Function
def get_wordnet_pos(tag):
    if tag.startswith("j"):
        return 'a'
    elif tag.startswith("R"):
        return 'a'
    elif tag.startswith("N"):
        return 'n'
    elif tag.startswith("A"):
        return 'a'
    elif tag.startswith("V"):
        return 'v'
    else:
        return 'n'
```
