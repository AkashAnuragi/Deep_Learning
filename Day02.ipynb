```python
# TEXT PREPROCESSING 
# 1. lowercase
st = "I love NLP part of Machine Learning"
st = st.lower()
st
```

```python
# 2. Tokenization => Convert Sentence into words
li = st.split()
```

```python
li
```

```python
# !pip install nltk
```

```python
from nltk.tokenize import word_tokenize
```

```python
st
```

```python
import nltk
nltk.download('punkt_tab')
```

```python
tokens = word_tokenize(st)
```

```python
tokens
```

```python
# 3. Remove Stop Words
nltk.download('stopwords')
```

```python
from nltk.corpus import stopwords
```

```python
stopword = set(stopwords.words("english"))
```

```python
len(stopword)
```

```python
new_tokens = [word for word in tokens if word not in stopword]
new_tokens
```

```python
new_tokens = []
for word in tokens:
    if word not in stopword:
        new_tokens.append(word)
new_tokens
```

```python
# 4. Stemming (convert your words to the base form)
# Disadvantage
# 1. it will not allow the grammer
# 2. it will remove the extra prefix/suffix mostly
```

```python
li = ['played','running', 'happiness', 'boys', 'studies']
```

```python
nltk.download('punkt')
```

```python
from nltk.stem import PorterStemmer
```

```python
stemmer = PorterStemmer()
```

```python
for word in li:
    print(word," => ",stemmer.stem(word))
```

```python
# stemmer use the context to convert word into a base form
```

```python
# Lemmatization (same as Stemmer)
# Lemmatization use (Lemma) context + grammer
```

```python
li = ['playing','studies','better', 'happiness']
```

```python
nltk.download('wordnet')
```

```python
nltk.download('omw-1.4')
```

```python
from nltk.stem import WordNetLemmatizer
```

```python
lemmatizer = WordNetLemmatizer()
for word in li:
    print(word," => ",lemmatizer.lemmatize(word))
```

```python
lemmatizer.lemmatize(li[0], pos='v')
```

```python
lemmatizer.lemmatize(li[0], pos='a')
```
